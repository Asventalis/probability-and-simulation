```python
from sympy import Matrix

P = Matrix([
    [0, 6, 0, 0, 0, 0, 0],
    [0, 1, 5, 0, 0, 0, 0],
    [0, 0, 2, 4, 0, 0, 0],
    [0, 0, 0, 3, 3, 0, 0],
    [0, 0, 0, 0, 4, 2, 0],
    [0, 0, 0, 0, 0, 5, 1],
    [0, 0, 0, 0, 0, 0, 6]
])/6
Q = P[0:6, 0:6]
I = Matrix.eye(6)
N = (I - Q).inv()
t = N * Matrix([1, 1, 1, 1, 1, 1])
t
```

```python
from sympy import Matrix, Rational

# do the thing above for generic n
def expected_steps(n):
    P = Matrix.zeros(n+1)
    for i in range(n):
        P[i, i] = Rational(i, n)
        P[i, i+1] = Rational(n-i, n)
    Q = P[0:n, 0:n]
    I = Matrix.eye(n)
    N = (I - Q).inv()
    t = N * Matrix([1]*n)
    return t[0]

expected_steps(5)
```
