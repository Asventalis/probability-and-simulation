```python
def factorial(n):
    """Calculate the factorial of n."""
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def binomial(n, k):
    """Calculate the binomial coefficient "n choose k"."""
    if k < 0 or k > n:
        return 0
    if k == 0 or k == n:
        return 1
    k = min(k, n - k)  # Take advantage of symmetry
    num = den = 1
    for i in range(k):
        num *= n - i
        den *= i + 1
    return num // den
```

```python
result = 1
for n in range(20,1,-2):
    result *= binomial(n, 2)
result //= factorial(10)
print(result)
```
