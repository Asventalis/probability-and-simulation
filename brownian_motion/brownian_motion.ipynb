```python
import numpy as np

def generate_geometric_brownian_motions(mu, sigma, X0 = 1, N=1, T=10, L=1000):
    """
    Generate N Geometric Brownian Motions (Xs) with parameters T and L.

    Parameters:
    N (int): Number of Xs to generate.
    T (float): Total time duration.
    L (int): Number of time steps.

    Returns:
    np.ndarray: An array of shape (N, L) containing the Xs.
    """
    dt = T / L
    t = np.linspace(0, T, L)
    
    # Generate random increments
    dW = np.random.normal(0, np.sqrt(dt), size=(N, L))
    
    # Initialize the X array
    X = np.zeros((N, L))
    
    # Set initial value
    X[:, 0] = X0
    
    # Generate the X paths
    for i in range(1, L):
        X[:, i] = X[:, i - 1] + X[:, i - 1] * mu * dt + X[:, i - 1] * sigma * dW[:, i - 1]
    
    return X
```

```python
# Generate Geometric Brownian Motions and plot them
import matplotlib.pyplot as plt

mu = 10
sigma = 20
X0 = 10
N = 100
T = 1
L = 1000

X = generate_geometric_brownian_motions(mu, sigma, X0=X0, N=N, T=T, L=L)

!matlab inline
plt.figure(figsize=(10, 6))
for i in range(N):
    # make it log in y
    plt.plot(X[i])
plt.title('Geometric Brownian Motions')
plt.xlabel('Time Steps')
plt.ylabel('Value')
plt.legend()
plt.show()
```
